```python
import pandas as pd
import numpy as np
from sklearn.model_selection import train_test_split, cross_val_score
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.pipeline import Pipeline
from sklearn.metrics import accuracy_score
```

# Antes que nada, generaremos un conjunto de datos arbitrario y random
Para esto utilizaremos funciones random y pandas para crear un data frame

```python
# Estableceré mi semilla en 2 y quiero 500 datos por carácteristica
np.random.seed(2)
n = 500

# Conjunto de datos para intentar predecir si lloverá mañana, como en algnos
# de los ejemplos dados en el pdf
df = pd.DataFrame({
    "temperatura": np.random.normal(35,5,n), #media = 30, como en sonora!
    "humedad": np.random.normal(30,100,n),
    "nubosidad": np.random.normal(0,9,n), #en internet dice que la nubosidad se mide entre 0 - 8
    "precipitaciones_mm": np.random.normal(0,22,n),
})

# La probabilidad de que llueva mañana, la podríamos simular como una suma
# del estado de diferentes eventos, como la temperatura y presión
# como el hecho de que sabemos que si hay nubes, es muy probable que llueva
# y aun más probable si hay mucha humedad, entonces:
prob_lluvia = (
    0.5 * (df["humedad"]/100)
    + 0.3 * (df["nubosidad"]/ 8)
    + 0.2 * (df["precipitaciones_mm"] > 0).astype(int) #nos interesa solo saber si ya llovió.
    + 0.1 * (df["temperatura"] - 25) / (40 - 25)
)

df['llovera_mañana'] = (prob_lluvia > 0.5).astype(int)

print(df.head())
```

# Crearé un pipeline con algunos errores

```python
# escalamos antes de separar el conjunto de datos


# Dividimos el data frame
X = df.drop(columns=['llovera_mañana'])
y = df['llovera_mañana']

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

pipeline = Pipeline([
    ('scaler', StandardScaler()),
    ('model', LogisticRegression())
])

pipeline.fit(X_train, y_train)
y_pred = pipeline.predict(X_test)
print("El score obtenido: ",accuracy_score(y_test, y_pred))
```

Vemos que el score obtenido es muy bueno, veremos que ocurre si lo corregimos:
# Pipeline corregido

```python
# Dividimos el data frame antes de escalar
X2 = df.drop(columns=['llovera_mañana'])
y2 = df['llovera_mañana']

X2_train, X2_test, y2_train, y2_test = train_test_split(X2, y2, test_size=0.3, random_state=42)

# Creamos el pipeline
pipeline = Pipeline([
    ('scaler', StandardScaler()),
    ('model', LogisticRegression())
])

pipeline.fit(X2_train, y2_train)
y2_pred = pipeline.predict(X2_test)
print("El score obtenido: ",accuracy_score(y2_test, y2_pred))
```
